```python
import os
import sys
import time
import copy
sys.path.append('../SLM/')

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim

import numpy as np
import json
import math
import random

import fnmatch
import re
from itertools import product, groupby

import matplotlib
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

from tokenizers import Tokenizer

import models, init, train, measures
import importlib
# Reload the modules
importlib.reload( models)

device = 'cuda' if torch.cuda.is_available() else 'mps' if torch.backends.mps.is_available() else 'cpu'
print(device)

def collect(path, prefix, *keys):

    all_entries = os.listdir(path)
    matching_entries = []
    for entry in all_entries:

        if fnmatch.fnmatch(entry, prefix+'*'):

            check_keys = 1
            for key in keys:
                check_keys *= fnmatch.fnmatch(entry, '*'+key+'*')
            if check_keys:
                ending = re.sub(prefix, '', entry)
#                 params = re.findall(r'_P(\d+)', ending)
#                 assert params, f'{entry}: trainsize p not found!'
                array = re.findall(r'_(\d+)', ending)
                matching_entries.append({
                    'ending': ending,
#                     'P': int(params[0]),
                    'seed': int(array[0])
                })
#     matching_entries.sort(key=lambda x: [x.get('P'), x.get('exp')])
    return matching_entries


def find_all(path, prefix, what, *keys):
    
    all_entries = os.listdir(path)
    params = []
    for entry in all_entries:

        if fnmatch.fnmatch(entry, prefix+'*'):

            check_keys = 1
            for key in keys:
                check_keys *= fnmatch.fnmatch(entry, '*'+key+'*')
            if check_keys:
                ending = re.sub(prefix, '', entry)
                param = re.findall(what+r'(\d+)', ending)
                param = int(param[0])
                if param not in params:
                    params.append(param)
    params.sort()
    return params


def find_lowest(path, prefix, what, *keys):
    
    all_entries = os.listdir(path)
    params = []
    for entry in all_entries:

        if fnmatch.fnmatch(entry, prefix+'*'):

            check_keys = 1
            for key in keys:
                check_keys *= fnmatch.fnmatch(entry, '*'+key+'*')
            if check_keys:
                ending = re.sub(prefix, '', entry)
                param = re.findall(what+r'(\d+)', ending)
                param = int(param[0])
                if param not in params:
                    params.append(param)
    params.sort()
    return params[0]


def get_loss(path, prefix):

    loss_list = []

    p_list = find_all(path, prefix, 'P')
    print('training set sizes: ', p_list)

    for  p in p_list:

        bs = find_lowest(path, prefix, 'bs', f'P{p}_')
        print(f'P={p}, batch size={bs}')
        collection = collect( path, prefix, f'P{p}_', f'bs{bs}_')
        loss = []

        for entry in collection:
            try:
                results = torch.load(path+prefix+entry['ending'])
                print(p, results['best']['loss'])
                loss.append(results['best']['loss'])

            except:
                print(path+prefix+entry['ending']+f' not found.')


        print(f'{len(loss)} exps. , mean loss {torch.tensor(loss).mean()}')
        loss_list.append(torch.tensor(loss))
    return loss_list


fig, ax = plt.subplots(1, 2, figsize=(20,5.0))
```

# $T=64$

## $n$-grams collapse

```python
matplotlib.use("pgf")
matplotlib.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'serif',
    'text.usetex': True,
    'pgf.rcfonts': False,
    'font.size': 10,
    'lines.markersize': 4,
    'lines.linewidth': 1.0,
})
fig = plt.figure(constrained_layout=True, figsize=(2*2.1, 1*2.2), dpi=120)

# example: 2 columns of plots + a colorbar column
gs = fig.add_gridspec(nrows=1, ncols=3, width_ratios=[1, 1, 0.05], wspace=0.02)
ax = np.array([
    fig.add_subplot(gs[0]),
    fig.add_subplot(gs[1]),
])

T = 64
depth = 12

gamma = 0.34
beta = 0.88

results = torch.load(f"tinystories/learning-curves/ngrams_tinystories.BPE8192_T{T}_d{depth}.pt")
print(len(results['ngrams']), results['ngrams'][0].size())
p_list = results['data']
ngrams_list = [n.mean(dim=0) for n in results['ngrams']]

base = matplotlib.colormaps["plasma"]                  # purple -> yellow
cmap  = matplotlib.colors.LinearSegmentedColormap.from_list(
    "plasma_trunc", base(np.linspace(0.05, 0.9, 256))
)
norm = matplotlib.colors.LogNorm(vmin=1, vmax=128+1)


for t in range(min(128,T)):

    ax[0].plot([p for p in p_list], [loss[t] for loss in ngrams_list], color=cmap(norm(t+1)), marker='.', markeredgecolor="black", markeredgewidth=0.2, label=f'{t+2}-gram')

    scaling = (t+1)**(2*beta)
    ax[1].plot([p/scaling for p in p_list], [loss[t]*(t+1)**gamma for loss in ngrams_list], color=cmap(norm(t+1)), marker='.', markeredgecolor="black", markeredgewidth=0.2, label=f'{t+2}-gram')

ax[0].set_xscale('log')
ax[0].set_yscale('log')

ax[1].set_xscale('log')
ax[1].set_yscale('log')


ax[0].set_ylabel(r'n-gram loss $\mathcal{L}_{n}$', fontsize=10)
ax[0].set_xlabel(r'training set size $P$', fontsize=10)
ax[1].set_ylabel(r'$\mathcal{L}_{n}\times n^\gamma$', fontsize=10, labelpad=-10)
ax[1].set_xlabel(r'$P/n^{2\beta}$', fontsize=10)

import matplotlib.ticker as mticker
ticks  = [ 2, 3, 4]
labels = [ "2", "3", "4"]  # must match length/order
ax[0].set_yticks(ticks)
ax[0].set_yticklabels(labels)
ax[0].yaxis.set_minor_locator(mticker.NullLocator())

ticks  = [4, 16]
labels = [ "4", "16"]  # must match length/order
ax[1].set_yticks(ticks)
ax[1].set_yticklabels(labels)
ax[1].yaxis.set_minor_locator(mticker.NullLocator())

sm = matplotlib.cm.ScalarMappable(norm=norm, cmap=cmap)
sm.set_array([])  # optional, keeps some mpl versions happy

cax = fig.add_subplot(gs[2])  # spans both rows -> matches plot stack height
cbar = fig.colorbar(sm, cax=cax)
cbar.ax.set_title(r"$n$", fontsize=10, pad=6)


ticks = 2 ** np.arange(
    int(np.floor(np.log2(1))),
    int(np.ceil(np.log2(128))) + 1
)
cbar.set_ticks(ticks)
exps = np.log2(ticks).astype(int)
cbar.set_ticklabels([rf"$2^{{{e}}}$" for e in exps])
```

```python
fig.savefig('tinystories/'+f'ngrams-scaling_tinystories.BPE8192_T64_d12.pdf', bbox_inches="tight")
```

```python
## fitting
import scipy
from scipy.optimize import curve_fit

def power_law_floor(x, A, alpha, C):
    return C + A * x**alpha

T = 128
depth = 12
results = torch.load(f"tinystories/learning-curves/learning-curve_tinystories.BPE8192_T{T}_d{depth}.pt")

p_list = results['data']
xdata = np.array(p_list)
error_list = results['losses']
ydata = np.array([loss.mean() for loss in error_list])

# 0.88 -0.27105540787055316 4.420242013541939 0.9999589063336854
entropy = 0.88
alpha = 0.27105540787055316
A = math.exp(4.420242013541939)

matplotlib.use("pgf")
matplotlib.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'serif',
    'text.usetex': True,
    'pgf.rcfonts': False,
    'font.size': 10,
    'lines.markersize': 4,
    'lines.linewidth': 1.0,
})
fig, ax = plt.subplots(1, 2, figsize=(2*3.8, 1*2.4), dpi=120)

gamma = 0.34
beta = 0.88

ax[0].errorbar([p for p in p_list], [loss.mean() for loss in error_list], yerr=[loss.std() for loss in error_list], color=f'C1', marker='.', label=fr'$T={T}$')
ax[0].plot([2**18,2**29],[38*(2**18)**-(gamma/(2*beta)),38*(2**29)**-(gamma/(2*beta))],'k--', label=r'$P^{-\gamma/(2\beta)}$')
ax[1].plot([p for p in p_list], [loss.mean()-floor for loss in error_list], color=f'C1', marker='.', label=r'$H_{T}=$'+'{:04.3f}'.format(entropy))

ax[1].plot([2**18,2**29],[30*(2**18)**-(gamma/(2*beta)),30*(2**29)**-(gamma/(2*beta))],'k--', label=r'$P^{-\gamma/(2\beta)}$')
ax[1].plot([2**18,2**29],[A*(2**18)**-(alpha),A*(2**29)**-(alpha)],'r--')
ax[0].plot(p_list, [A*p**(-alpha)+entropy for p in p_list], 'C1', linestyle='dashed', label=r'$\alpha_P=$'+'{:04.3f}'.format(alpha))


ax[0].legend(fontsize=8, handlelength=1)
ax[1].legend(fontsize=8, handlelength=1)
# ax.set_xlim(1,16)
ax[0].set_xscale('log')
# ax.set_ylim(1.,5.)
ax[0].set_yscale('log')

ax[1].set_xscale('log')
ax[1].set_yscale('log')
ax[1].set_xlim(1e5,1e9)
# ax.set_ylim(1.,5.)

ax[0].set_ylabel(fr'AR loss', fontsize=11)
ax[0].set_xlabel(fr'training set size $P$', fontsize=11)

ax[1].set_ylabel(r'$\mathcal{L}(P)-\mathcal{L}(\infty)$', fontsize=11)
ax[1].set_xlabel(r'$P$', fontsize=11)

fig.tight_layout()
```

```python
matplotlib.use("pgf")
matplotlib.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'serif',
    'text.usetex': True,
    'pgf.rcfonts': False,
    'font.size': 10,
    'lines.markersize': 4,
    'lines.linewidth': 1.0,
})
# fig, ax = plt.subplots(1, 1, figsize=(1*3.8, 1*2.4), dpi=120)
fig, ax = plt.subplots(1, 2, figsize=(2*3.8, 1*3.4), dpi=120)


T = 128
depth = 12

results = torch.load(f"tinystories/learning-curves/ngrams_tinystories.BPE8192_T{T}_d{depth}.pt")
print(len(results['ngrams']), results['ngrams'][0].size())
p_list = results['data']
xdata = np.array(p_list)
ngrams_list = [n.mean(dim=0) for n in results['ngrams']]

t = 0
ax[0].plot([p for p in p_list], [loss[t] for loss in ngrams_list], color=f'C{t}', marker='.', label=f'n={t+1}')
# ax[1].plot([p for p in p_list], [loss[t]-A*(t+1)**(-gamma) for loss in ngrams_list], color=f'C{t}', marker='.')

for i, floor in enumerate([3.53,3.54,3.55,3.56,3.57]):

    ax[1].plot([p for p in p_list], [loss[t]-floor for loss in ngrams_list], color=f'C{i}', marker='.')
    ydata = np.array([loss[t]-floor for loss in ngrams_list])
    slope, intercept, r_value, p_value, std_err = scipy.stats.linregress(np.log(xdata[2:]), np.log(ydata[2:]))
    r2 = r_value**2
    print(t+1,floor, slope, intercept, r2)
    ax[1].plot([p for p in p_list], [loss[t]-floor for loss in ngrams_list], color=f'C{i}', marker='.', label=fr'$H_1={floor}$')#, $\alpha={slope}$, $r^2={r2}$')

ax[1].plot([1e6,1e9],[10*1e6**-(.33/(2*.872)),10*1e9**-(.33/(2*.872))],'r--')
# 1 3.55 -0.32712957893373273 3.827039289474487 0.9920340012017484
A = math.exp(3.827039289474487)
C = 3.55
alpha = 0.32712957893373273
ax[1].plot([1e6,1e9],[A*1e6**-(alpha),A*1e9**-(alpha)],'k--', label=r'$\alpha=$'+'{:04.3f}'.format(alpha))
ax[0].plot(p_list, [A*p**(-alpha)+C for p in p_list], f'C{t}', linestyle='dashed', label=fr'$H_1={C}$')

ax[0].legend(fontsize=8, handlelength=1)
ax[0].set_xscale('log')
ax[0].set_yscale('log')

ax[1].legend(fontsize=8, handlelength=1)
ax[1].set_xscale('log')
ax[1].set_yscale('log')

ax[0].set_ylabel(r'$\mathcal{L}_{1}$', fontsize=11)
ax[0].set_xlabel(r'training set size $P$', fontsize=11)
ax[1].set_ylabel(r'$\mathcal{L}_{1}-H_1$', fontsize=11)
ax[1].set_xlabel(r'$P$', fontsize=11)

fig.tight_layout()
```
