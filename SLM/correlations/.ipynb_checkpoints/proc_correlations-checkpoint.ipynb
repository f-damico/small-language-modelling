```python
import os
import sys
import time
import copy
sys.path.append('..')

import numpy as np
import math
import random

import torch
import torch.nn as nn
import torch.nn.functional as F

import json
```

```python
%%time

vocab_size = 8192
split = 'valid'
path = '../datasets/tinystories/'
filename = f'tinystories.BPE{vocab_size}.{split}.npy'

device = 'cuda' if torch.cuda.is_available() else 'mps' if torch.backends.mps.is_available() else 'cpu'

corpus = np.load(path+filename)
corpus = torch.tensor(corpus).to(device)
corpus_len = corpus.size(0)

max_dist = 1024
dist = 64
assert dist < max_dist, f'dist must be smaller than max_dist={max_dist}!'

num_data = 2**20
batch_size = 2**15

if batch_size is None:
    batch_size = num_data
num_batches = num_data//batch_size
samples = torch.tensor( random.sample( range(max_dist-1,corpus_len), num_data))
print(samples.size())
print('total dataset size: ', num_data, f'(corpus length = {corpus_len}), split into {num_batches} batches.')
```

```python
%%time
cov = torch.zeros( # right_feature x left_feature
    (vocab_size, vocab_size), device=device
)
prob_right = torch.zeros(
    (vocab_size), device=device
)
prob_left = torch.zeros(
    (vocab_size), device=device
)
ckpt = 1

cov_buf = torch.empty_like(cov, device='cpu', pin_memory=True)
right_buf = torch.empty_like(prob_right, device='cpu', pin_memory=True)
left_buf = torch.empty_like(prob_left, device='cpu', pin_memory=True)

for i in range(num_batches):

    samples_batch = samples[batch_size*i:batch_size*(i+1)]

    right = F.one_hot( corpus[samples_batch].long(), num_classes=vocab_size).float()
    prob_right += right.sum(dim=0)

    left = F.one_hot( corpus[samples_batch-dist].long(), num_classes=vocab_size).float()
    prob_left += left.sum(dim=0)

    cov += right.t() @ left

    if (i+1)==ckpt:

#         corr = (cov/(ckpt*batch_size)-(prob_right[:,None]/(ckpt*batch_size))@(prob_left[None,:]/(ckpt*batch_size))).std()
#         print('saving at ', ckpt*batch_size, f' data, corr={corr}')

        corr = (cov/(ckpt*batch_size)-(prob_right[:,None]/(ckpt*batch_size))@(prob_left[None,:]/(ckpt*batch_size))).std().cpu()
        right_buf.copy_(prob_right, non_blocking=True)
        left_buf.copy_(prob_left, non_blocking=True)
        cov_buf.copy_(cov, non_blocking=True)

        print('saving at ', ckpt*batch_size, ' data')
        torch.save(
            {'P': ckpt*batch_size, 'right': right_buf/(ckpt*batch_size), 'left': left_buf/(ckpt*batch_size), 'cov': (cov_buf/(ckpt*batch_size)), 'corr': corr},
            f"testcorr_P{ckpt*batch_size}.pt"
        )
        ckpt *= 2
```

```python
P = 32768
results = torch.load(f"testcorr_P{P}.pt")
print(results.keys())

esempio1 = results['right']
print(esempio1.size(), esempio1.device)

esempio2 = results['cov']
print(esempio2.size(), esempio2.device)

esempio3 = results['corr']
print(esempio3, esempio3.device)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 1, figsize=(6.5,5))

x_pts = [2**15, 2**16, 2**17, 2**18, 2**19, 2**20]
y_pts = []
for x in x_pts:
    results = torch.load(f"testcorr_P{x}.pt")
    y_pts.append(results['corr'].cpu())

ax.plot(x_pts, y_pts)
ax.plot(x_pts, [x**(-.5)/(vocab_size) for x in x_pts], 'k--')

ax.set_xscale('log')
ax.set_yscale('log')
```
